### （１）非言語性の推論能力検査

知的障害の疑いがある児童を解析対象から除外するために、簡便な知能検査として用いられるレーヴン色彩マトリックス検査（Raven，1962; 杉下・山崎，1993）を集団式で行った。各学年の教室内に設置されている電子黒板を使用して各問題の刺激をスクリーンに呈示した。子どもはスクリーン上の刺激を見て、回答用紙に記載された選択肢の数字に丸をつけた。

### （２）漢字の読み能力検査

対象児童の漢字の読み能力を測るために標準読み書きスクリーニング検査（STRAW-R： 宇野ら,　2017）の漢字126語音読課題を実施した。検査は集団式で行い、対象児童は呈示された漢字刺激に対してひらがなで読み方を書いて回答した。集団式検査での実施に伴い、以下の３つの点を変更した。

1. 126 語を ５ つのセット（26 語 １ セット、25 語 ４ セット）に分けそれぞれのセットを A４ 用紙 １ 枚に印刷して呈示した。また、習っていない漢字単語が続くと対象児童の課題に取り組む意欲が下がってしまうことを考慮し、単語の提示順を入れ替えた。それぞれのセットに必ず配当学年 １〜６ 年生の漢字が存在するようにした。
2. 回答について、対象児童にひらがなで回答を求めたが、漢字を正しく読めることについて焦点を当てたため、ひらがなの字形の誤り（例：「か」の 3 画目がない）、長音の誤り（例：めいじん→めえじん）は正答とした。
3.  STRAW-R のマニュアルでは、単語としては正しい読み方ではないが、構成文字をその文字がもつ他の読み方で読んでいる場合（例:「競馬」を「きょうば」と読む）には、他の読み方を尋ねるように記載されている。しかし、集団式検査としたため、他の読み方を尋ねることはできないことを理由に、本研究では行わなかった。

### （３）漢字実在語の音読課題
漢字二字熟語の音読成績における一貫性効果を検討するために、漢字実在語の音読課題を実施した。刺激は、漢字二字の典型語と非典型語とした。刺激数は、３ 年生では小学 １・２ 年で習う漢字からなる 50 語（典型語 25 語、非典型語 25 語）、４ 年生では ３ 年生と同一の刺激に加え、３ 年生で習う漢字を含む語からなる 50 語（典型語 25 語、非典型語 25 語）の計 100 語、５・６ 年生では ４ 年生と同一の刺激に加え、４ 年生で習う漢字を含む語から成る 40 語（典型語 20 語、非典型語 20 語）の計 140 語とした。これらの刺激は、以下のように選定されている。


まず、刺激選定用のコーパスを作成した。NTT データベース「日本語の語彙特性」(1999) に掲載されている漢字二字の熟語のうち、① 音声文字呈示の親密度、音声呈示の親密度、文字呈示の親密度の値が全て ５ 以上であること、② 頻度値、音声呈示の心像性、文字呈示の心像性の値すべてが掲載されていることの ２ 条件を満たす単語を抽出した。そして、それらの単語の各構成文字の配当学年を調べ、次の ４ つのセット、① 配当学年が 2 年生までの字で構成されている漢字二字熟語（配当学年が ２ までの単語）、② 少なくとも片方の文字の配当学年が 3 年で、もう片方の文字の配当学年が 1～3 年の漢字二字熟語（配当学年が 3 年の単語）、③ 少なくとも片方の文字の配当学年が4年で、もう片方の文字の配当学年が 1～4 年の漢字二字熟語（配当学年が 4 年の単語）、④ 少なくとも片方の文字の配当学年が 5 年で、もう片方の文字の配当学年が 1～5 年の漢字二字熟語（配当学年が5年の単語）のセットを作成した。なお 3～6 年共通の刺激 50 語はセット ① から、4～6 年共通の刺激 50 語はセット ② から、5・6 年共通の刺激はセット ③ から、以下のように各単語を典型語と非典型語に分類して選定した。


各単語を典型語・非典型語に分類するために、Fushimi et al.（1999）における漢字の読みの一貫性値に関する定義に基づき、上記 ①～④ の単語セットを使用して、文字位置別に読みの一貫性値（Friend 数/隣接語数）を算出した。なお、本研究では各学年が知っていると思われる語から、Friend 数と隣接語数を算出したうえで、一貫性値を算出している。具体的には、3 年生の Friend 数と隣接語数をセット ① の単語（3 年生一貫性値算出用コーパス）から、4 年生のFriend 数と隣接語数をセット ① ② の単語（4 年生一貫性値算出用コーパス）から、5 年生の Friend 数と隣接語数をセット ①②③ の単語（5 年生一貫性値算出用コーパス）から、6 年生の Friend 数と隣接語数をセット ①②③④ の単語（6 年生一貫性値算出用コーパス）から算出し、各学年の一貫性値を算出した。そして、各学年の一貫性値に基づき、それぞれの文字の読み方を、その学年での典型的な読みと非典型的な読みに分類した。具体的には、各学年一貫性値算出用コーパス内で、文字位置別にみた際に、ある漢字に対する最も多い読み方（多数派の読み）で、かつ、その読み方の一貫性値が 0.6 以上の読み方を典型的な読みとし、多数派の読みではなく、その漢字の読みの一貫性値が 0.4 以下の読み方を非典型的とした。さらに、それぞれの読み方の一貫性（典型的、非典型的）に基づき、それぞれの漢字二字熟語を、その学年での典型語、非典型語に分類した。本研究では、両文字の読み方が典型的な漢字二字熟語を典型語とし、少なくとも一方の漢字の読み方が非典型的な漢字二字熟語を非典型語とした。


３～６ 年生共通の音読刺激として、どちらの構成漢字も配当学年が ２ 以下で、かつ、一文字目、二文字目いずれも隣接語数が ４ 以上の単語 471 語の中から、どの学年にとっても典型語に分類されていた 25 語と、非典型語に分類されていた 25 語を選択した。４～６ 年生共通の音読刺激として、どちらの構成漢字も配当学年が ３ 以下で、かつ、一文字目、二文字目いずれも隣接語数が ４ 以上の単語 445 語の中から、どの学年にとっても典型語に分類されていた 25 語と、非典型語に分類されていた 25 語を選択した。そして、５・６ 年生共通の音読刺激として、一文字目、二文字目のうち少なくとも一方の構成漢字配当学年が ４ で、かつ、一文字目、二文字目いずれも隣接語数が ４ 以上の単語 410 語から、どちらの学年にとっても典型語に分類されていた 20 語と、非典型語に分類されていた20 語を選択した。典型語条件と非典型語条件の間で、NTT データペースシリーズ「日本語の語彙特性」から得た文字親密度、 log 頻度、文字心像性の値を統制した。Table１ に用いた刺激における属性値の平均、SD、t検定の結果を示す。
